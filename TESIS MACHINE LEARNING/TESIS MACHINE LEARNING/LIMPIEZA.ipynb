```python
def limpiar_espacios(df):
    columnas = df.select_dtypes(include="object").columns
    for c in columnas:
        df[c] = (
            df[c]
            .astype(str)
            .str.strip()
        )
    return df
```

```python
ensanut = limpiar_espacios(ensanut)
endi = limpiar_espacios(endi)
```

```python
VALORES_NULOS = [""," ",".","-","--","---","NA","N/A","Na","Nan","nan","NULL"]
```

```python
def convertir_nan(df):
    return df.replace(
        VALORES_NULOS,
        np.nan
    )
```

```python
ensanut = convertir_nan(ensanut)
endi = convertir_nan(endi)
```

```python
# ==========================================================
# NORMALIZAR CATEGORÍAS
# ==========================================================

def normalizar_categoria(valor):
    if pd.isna(valor):
        return valor
    if not isinstance(valor,str):
        return valor
    valor = valor.strip()
    valor = valor.title()
    reemplazos = {
        "Si":"Si",
        "si":"Si",
        "no":"No",
        "No Se":"No",
        "No Sabe":"No sabe",
        "no sabe":"No sabe",
        "no sabe / no responde": "No sabe/No responde",
        "No Responde":"No responde",
        "NS/NR" : "No sabe/No responde",
        "Ns/Nr" : "No sabe/No responde",
        "ns/nr" : "No sabe/No responde"
    }
    return reemplazos.get(valor,valor)
```

```python
for c in ensanut.select_dtypes(include="object"):
    ensanut[c] = ensanut[c].apply(normalizar_categoria)

for c in endi.select_dtypes(include="object"):
    endi[c] = endi[c].apply(normalizar_categoria)
```

```python
def perfil_variables(df):
    perfil = pd.DataFrame({
        "Variable": df.columns,
        "Tipo": df.dtypes.astype(str),
        "Registros": len(df),
        "No nulos": df.notna().sum().values,
        "Nulos": df.isna().sum().values,
        "% Nulos": np.round(df.isna().mean().values * 100, 2),
        "Valores únicos": df.nunique(dropna=True).values
    })
    return perfil
```

```python
perfil_ensanut = perfil_variables(ensanut)
perfil_endi = perfil_variables(endi)

display(perfil_ensanut.head())
display(perfil_endi.head())
```

```python
def clasificar_variable(nombre, serie):
    nombre = nombre.lower()
    if pd.api.types.is_numeric_dtype(serie):
        return "Numérica"
    return "Categórica"
```
