```python
# ==========================================================
# OUTLIERS IQR ENSANUT
# ==========================================================

reporte_outliers=[]
for col in ensanut.select_dtypes(include=["number"]).columns:
    Q1=ensanut[col].quantile(0.25)
    Q3=ensanut[col].quantile(0.75)
    IQR=Q3-Q1
    inferior=Q1-1.5*IQR
    superior=Q3+1.5*IQR
    cantidad=((ensanut[col]<inferior)|
               (ensanut[col]>superior)).sum()
    porcentaje=round(cantidad/len(ensanut)*100,2)
    reporte_outliers.append({
        "Variable":col,
        "Outliers":cantidad,
        "%":porcentaje
    })
reporte_outliers=pd.DataFrame(reporte_outliers)
display(
reporte_outliers.sort_values(
    "%",
    ascending=False
)
)
```

```python
# ==========================================================
# OUTLIERS IQR ENDI
# ==========================================================

reporte_outliers=[]
for col in endi.select_dtypes(include=["number"]).columns:
    Q1=endi[col].quantile(0.25)
    Q3=endi[col].quantile(0.75)
    IQR=Q3-Q1
    inferior=Q1-1.5*IQR
    superior=Q3+1.5*IQR
    cantidad=((endi[col]<inferior)|
               (endi[col]>superior)).sum()
    porcentaje=round(cantidad/len(endi)*100,2)
    reporte_outliers.append({
        "Variable":col,
        "Outliers":cantidad,
        "%":porcentaje
    })
reporte_outliers=pd.DataFrame(reporte_outliers)
display(
reporte_outliers.sort_values(
    "%",
    ascending=False
)
)
```

```python
from scipy.stats import zscore
reporte=[]
for col in ensanut.select_dtypes(include=["number"]).columns:
    datos=ensanut[col].dropna()
    if datos.nunique()>1:
        z=np.abs(zscore(datos))
        cantidad=(z>3).sum()
        porcentaje=round(cantidad/len(datos)*100,2)
        reporte.append({
            "Variable":col,
            "Outliers_Z":cantidad,
            "%":porcentaje
        })
display(
pd.DataFrame(reporte).sort_values(
    "%",
    ascending=False
)
)
```

La presente investigación emplea dos fuentes oficiales de información: la Encuesta Nacional de Salud y Nutrición (ENSANUT 2018) y la Encuesta Nacional sobre Desarrollo Infantil (ENDI). Sin embargo, únicamente la base de datos ENSANUT será utilizada para el desarrollo y entrenamiento de los modelos de Machine Learning, debido a que contiene la variable objetivo correspondiente al estado nutricional infantil (desnutrición crónica), además de variables antropométricas, demográficas, socioeconómicas y de atención en salud que permiten construir modelos predictivos supervisados. Por su parte, la base ENDI constituye una fuente complementaria para el análisis descriptivo del contexto del desarrollo infantil temprano, proporcionando información sobre estimulación, cuidado, educación inicial y programas sociales que enriquecen la interpretación de los resultados, aunque no puede incorporarse directamente al entrenamiento del modelo al no contener la variable objetivo ni compartir identificadores comunes con ENSANUT. De esta manera, ambas bases de datos se aprovechan de forma metodológicamente adecuada, preservando la validez del análisis predictivo y fortaleciendo la discusión de los factores asociados al estado nutricional de la población infantil de Tungurahua.
